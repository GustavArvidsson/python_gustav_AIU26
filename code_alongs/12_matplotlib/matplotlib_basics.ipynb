## matplotlib


```python
import numpy as np 

x = np.linspace(-2*np.pi, 2*np.pi)


```

```python
y = np.sin(x)
```

## MATLAB style interface - dont do this

```python
import matplotlib.pyplot as plt

plt.figure(dpi=100, figsize=(12,4))

plt.plot(x,y)
plt.title("sin(x)")
plt.xlabel("X")
plt.ylabel("y")
```

## OOp approach

```python
fig, ax = plt.subplots(1)
ax.plot(x,y)
ax.set(title="sin(x)", xlabel="x", ylabel="y")
```

## subplots

```python
fig, ax = plt.subplots(1,2, figsize=(16,4))

ax[0].plot(x, np.sin(x))
ax[0].plot(x, np.cos(x))
ax[0].legend(["sin x", "cos x"])


ax[1].plot(x, np.exp(x))
ax[1].set(title="Exponetial function")

fig.savefig("cool_graph.png", dpi=250)
```
