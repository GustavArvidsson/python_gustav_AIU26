```python
user_input = input("What is your room number?")

floors = [1, 2, 3]

def room_to_floor(user_input):
    if len(user_input) != 3:
        print("You need to enter 3 numbers")
    elif user_input[0] == "1":
        print(f"Your room is osdhjyn the {floors[0]}st floor")
    elif user_input[0] == "2":
        print(f"Your room is on the {floors[1]}nd floor")
    elif user_input[0] == "3":
        print(f"Your room is on the {floors[2]}rd floor")
    else:
        print("That room number doesn't correspond to any floor. We have 3 floors, the first number in the room number corresponds to one of floors")


room_to_floor(user_input)
```

```python
def get_floor(floor_number):
    try:
        floor = floor_number[0]

        if floor == "1":
            print("You are on floor 1.")
        elif floor == "2":
            print("You are on floor 2.")
        elif floor == "3":
            print("You are on floor 3")
    except ValueError as err:
        print(err)


get_floor(str(307))
```

## arbitrary arguments * args
- can take in arbitrary number of arguments
- all arguments goes into tuple

```python
def check_args(*numbers):
    print(numbers)
    print(numbers[4:])
    print(len(numbers))

check_args(1, 2, 3, 4, 5, 6, 1)
```

```python
def mean(*numbers):
    sum_ = 0
    for number in numbers:
        sum_+= number
    return sum_ / len(numbers)

mean(1, 2, 6), mean(2, 5, 6), mean(4, 4, 3), mean(1)
```

## **kwargs - arbitrary keyword arguments
- can have as many as possible keyword arguments
- the keyword arguments becomes a dictionary

```python
def test_kwargs(**kwargs):
    print(kwargs)

    for key, value in kwargs.items():
        print(f"{key}: {value}")

test_kwargs(greet="Hej", cool_number=15, is_cool=True)
```

ex - wrapper function

```python
import matplotlib.pyplot as plt

x = range(5)
y = [2, -3, 5, 2, 8]

plt.plot(x, y)
plt.grid()
plt.title("Random stock")
plt.xlabel(xlabel="days")
plt.ylabel(ylabel="price")
```

```python

x = range(5)
y = [-5, -30, 5, 20, 8]

plt.plot(x, y, "o--")
plt.grid()
plt.title("cooler stock")
plt.xlabel(xlabel="days")
plt.ylabel(ylabel="price")
```

```python
x = range(5)
y = [-5, -30, 5, 20, 8]

# det är ** som ger att man kan skicka in ett godtyckligt antal argument som nyckelord.
def stock_plot(x, y, title, **options):
    plt.plot(x, y, **options)
    plt.grid()
    plt.title(label=title)
    plt.xlabel(xlabel="days")
    plt.ylabel(ylabel="price")

stock_plot(x, y, title="Cooler stock", linewidth=4, linestyle="--", marker="o", color="orange")
```

```python
#easy way to call the function with keyword arguments
stock_plot(x, y, title="Cooler stock", linewidth=4, linestyle="--", marker="o", color="green")
```
