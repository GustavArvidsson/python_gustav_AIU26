## Diictionary

```python
person = dict(
    name="Gustav",
    age=22,
    work="studying",
    interests=["AI", "Filming", "Outdoors"],
    phone=112,
)

person
```

```python
person ={
    'name': 'Gustav',
 'age': 22,
 'work': 'studying',
 'interests': ['AI', 'Filming', 'Outdoors'],
 'phone': 112
}

person
```

```python
print(f"{person['name']}")
print(f"{person['age']}")
print(f"{person['work']}")
print(f"{person['interests'][1] = }")
```

## looping through a dict

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```
