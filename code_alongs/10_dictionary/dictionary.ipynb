## Diictionary

```python
person = dict(
    name="Gustav",
    age=22,
    work="studying",
    interests=["AI", "Filming", "Outdoors"],
    phone=112,
)

person
```

```python
person ={
    'name': 'Gustav',
 'age': 22,
 'work': 'studying',
 'interests': ['AI', 'Filming', 'Outdoors'],
 'phone': 112
}

person
```

```python
print(f"{person['name']}")
print(f"{person['age']}")
print(f"{person['work']}")
print(f"{person['interests'][1] = }")
```

## looping through a dict

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```

```python
person["has_driver_license"] = True
person
```

```python
# F<10, E: 10, D:20, C:30, B:40, A:50
# empty dict
grade_limits={}

for i, char_ in enumerate("ABCDEF"[::-1]):
    # assigns new keys and values to the dict
    grade_limits[char_] = i*10

grade_limits["F"] = "<10"
grade_limits
```

```python
#Betygsgränser
# A: 50
# B: 40



for grade, limit in reversed(grade_limits.items()):
    print(f"{grade}: {limit}")
    
```

```python
# grades = "ABCDEF"
grades = sorted(grade_limits.keys())

for grade in grades:
    print(f"{grade}: {grade_limits[grade]}")
```

## dictionary comprehension

```python
grade_limits = {grade: i*10 for i, grade in enumerate("FEDCBA")}
grade_limits
```
