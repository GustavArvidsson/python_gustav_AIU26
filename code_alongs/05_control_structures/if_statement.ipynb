```python
# Balder
# Requierment 140cm and less than 190cm
# At least 12 years

height = int(input("Enter your height in cm: "))
age = int(input("Enter your age: "))

if height >= 140 and height <= 190 and age >= 12:
    print("Welcome to balder!")
else:
    print("You can go to kaninlandet instead!")
```

```python
# Exam results
# A: 90
# B: 80
# C: 70
# D: 60
# E: 50
# F: <50
#
# let user input a score

score = int(input("What score did you get? "))

if score >= 90:
    print("You got an A!")
elif score >= 80:
    print("You got a B!")
elif score >= 70:
    print("You got a C!")
elif score >= 60:
    print("You got a D!")
elif score >= 50:
    print("You got an E!")
else:
    print("You got an F!")
```

```python
# is my number even or odd?

number = int(input("Enter a number: "))

if number % 2 == 0: #means devisable by 2 with no remainder
    print("Even!")
elif number % 2 == 1: #if number has a remainder when divided by 2, it is odd
    print("Odd!")
```
