## random

count number of fives in 1 000 000 dice rolls -> calculate proportions of fives

```python
import random as rnd
fives = 0
for i in range(1,1000000):
    dice = rnd.randint(1,6)
    if dice == 5:
        fives += 1
print(fives)
```
