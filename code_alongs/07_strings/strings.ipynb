## concatinate strings

```python
firstname = "      bob"
lastname = "       dylan"

print(firstname + " " + lastname)
print(f"{firstname} {lastname}")
```

```python
greet = f"""
HELLO
{firstname.strip().capitalize()} {lastname.strip().capitalize()}
"""
print(greet)
```

## concatenate using loop

```python
bamba =["pancake", "pokebowl", "fish n chips", "potatisbullar", "pytt i panna"]
days = ["mån", "tis", "ons", " tor", "fre"]
menu = "veckomeny"
for day, food in zip(days, bamba):
  menu += f"{day}: {food}\n"
```
