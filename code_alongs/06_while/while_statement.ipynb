# while statement

- loops when condition is True
- "if-loop"

```python
i=0

while i < 5:
    print(f"hi number {i}")
    i += 1 #increments i in each iteration
```

```python
i = -10

while i <= 10:
    #**2 to the power of 2
    square = i**2
    print(square, end=" ")
    i += 1
```

```python
birds =80000
year = 0

while birds > 8000:
    print(year, birds, end=" ")
    birds = birds * 0.5
    year += 1
```
