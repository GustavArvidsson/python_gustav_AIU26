## lists

```python
cities =["göteborg", "varberg" , "falkenberg", "stockholm", "malmö", "halmstad"]
print(f"{cities[0] = }")
print(f"{cities[3] = }")
print(f"{cities[-1] = }")
print(f"{cities[-3] = }")
# a string "stockholm" is a list of characters
print(f"{cities[-3][0:5] = }")
```

```python
#slicing a list
print(f"{cities[1:4] = }")
print(f"{cities[3:7] = }")
print(f"{cities[3:] = }")
```

modify list

```python
#mutate list, går att ändra listan
cities[3] = "kungsbacka"
cities
```

sort list

```python
cities.sort()
cities
```

```python
numbers = [-1, -5, 5, 9, 3]
numbers.sort(reverse=True)

numbers
```

## loop through a list

```python
for city in cities:
    print(city)
```

```python
more_cities = ["stockholm", "örebro", "umeå", "luleå"]

for city in more_cities:
    cities.append(city)

cities
```

create a list of cubes between 0 - 10

```python
cubes = []

for i in range(11):
    cubes.append(i**3)

cubes
```

## list comprehension

```python
cubes = [i**3 for i in range(11)] 
cubes
```

```python
even_numbers_cubed = [i**3 for i in range (11) if i%2 == 0]
even_numbers_cubed
```
