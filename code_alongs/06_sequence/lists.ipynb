## lists

```python
cities =["göteborg", "varberg" , "falkenberg", "stockholm", "malmö", "halmstad"]
print(f"{cities[0] = }")
print(f"{cities[3] = }")
print(f"{cities[-1] = }")
print(f"{cities[-3] = }")
# a string "stockholm" is a list of characters
print(f"{cities[-3][0:5] = }")
```

```python
#slicing a list
print(f"{cities[1:4] = }")
print(f"{cities[3:7] = }")
print(f"{cities[3:] = }")
```

modify list

```python
#mutate list, går att ändra listan
cities[3] = "kungsbacka"
cities
```

sort list

```python
cities.sort()
cities
```
