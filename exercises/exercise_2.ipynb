# 2. Medicine

```python
#Adults & adolescents over 12 years and over 40kg: 1-2 pills
#Children 7-12 years and 26-40kg: 1/2-1 pill
# children 3-7 years and 15-25kg: 1/2 pill

age = int(input("Enter your age: "))
weight = float(input("Enter your weight in kg: "))

if age > 12 and weight > 40:
    print("Take 1-2 pills.")
elif 7 <= age <= 12 and 26 <= weight <= 40:
    print("Take 1/2-1 pill.")
elif 3 <= age <= 7 and 15 <= weight <= 25:
    print("Take 1/2 pill.")

```

# 3. Divisible

```python
# is the number a. even or odd
#b. divisible by 5 or not
#c. is devisible by 5 and odd

number = int(input("Enter a number: "))

# a. even or odd
if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

# b. divisible by 5 or not
if number % 5 == 0:
    print("The number is divisible by 5.")
else:
    print("The number is not divisible by 5.")

# c. divisible by 5 and odd
if number % 5 == 0 and number % 2 == 1:
    print("The number is divisible by 5 and odd.")
else:
    print("The number is not divisible by 5 and odd.")
```

# 4. Luggage size

```python
max_weight = 8
max_length = 55
max_width = 40
max_height = 23

in_weight = float(input("Enter the weight of your luggage in kg: "))
in_length = float(input("Enter the length of the luggage in cm: "))
in_width = float(input("Enter the width of the luggage in cm: "))
in_height = float(input("Enter the height of the luggage in cm: "))

if in_weight <= max_weight and in_length <= max_length and in_width <= max_width and in_height <= max_height:
    print("Your luggage meets the requirements.")
elif in_weight > max_weight:
    print("Your luggage exceeds the weight limit.")
elif in_length > max_length:
    print("Your luggage exceeds the length limit.")
elif in_width > max_width:
    print("Your luggage exceeds the width limit.")
elif in_height > max_height:
    print("Your luggage exceeds the height limit.")
```

# 5. count numbers

```python
#use a for statement to count from:
# a) -10 to 10 with one increment
# b) -10 to 10 with two increments

one_inc = -10
two_inc = -10
print("Counting from -10 to 10 with one increment:")
for i in range(one_inc, 11):
    print(i, end=" ")
print()  # New line after the first count
print("Counting from -10 to 10 with two increments:")
for i in range(two_inc, 11, 2):
    print(i, end=" ")
print()  # New line after the second count
```

# 6. Multiplication game

```python
#When the player has answerd and gotten the result. He should be asked if he wants to play again. If yes, the game should start over, if no, the program should end and thank the player for playing.

import random

while True:
    x = random.randint(1, 10)
    y = random.randint(1, 10)

    answer = input(f"What is the product of {x} and {y}? ")
    if answer.isdigit() and int(answer) == x * y:
        print("Correct!")
    else:
        print(f"Incorrect. The correct answer is {x * y}.")

    play_again = input("Do you want to play again? (yes/no) ")
    if play_again.lower() != "yes":
        print("Thank you for playing!")
        break
```

# 7. Multiplication table

```python
#Use for statement(s) to:
# a)   print out the 6th multiplication table from 0 to 10 (*)
# b)   let the user input the table, start and end of the table. (*)
# c)   print out a full multiplication table from 0 to 10. (**)
# Hint: Use nested for-loops, which means a for-loop in a for-loop. To format the table nicely, use f-string:
# print(f"{number :4}", end = "")

# a) 6th multiplication table from 0 to 10
print("a) 6th multiplication table:")
for i in range(0, 11):
    print(f"6 x {i} = {6 * i}")

# b) user-defined table, start and end
print("\nb) Custom multiplication table:")
table = int(input("Enter the multiplication table: "))
start = int(input("Enter the start value: "))
end = int(input("Enter the end value: "))

for i in range(start, end + 1):
    print(f"{table} x {i} = {table * i}")

# c) full multiplication table from 0 to 10
print("\nc) Full multiplication table from 0 to 10:")
for i in range(0, 11):
    for j in range(0, 11):
        print(f"{i * j:4}", end="")
    print()
```

# 9. Check convergence

```python
#a)
n = 20
total_sum = 0.0
i = 0

while i <= n:
  total_sum += 1 / (2**i)
  i += 1

print(f"Summa a (while, n={n}): {total_sum}")
```
