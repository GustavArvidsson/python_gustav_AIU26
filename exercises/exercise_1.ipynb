# 1. Classification accuracy

```python
correct = 300
total = 365

accuracy = correct / total

print(f"Accuracy: {accuracy:.3f}")
print(f"Träffsäkerhet i procent: {accuracy * 100:.1f}%")
```

# 2. Classification accuracy

```python
TP = 2
FP = 2
FN = 11
TN = 985

accuracy = (TP + TN) / (TP + FP + FN + TN)

print(f"accuracy: {accuracy * 100:.1f}%")

# A quite bad model since it has 11 false negatives wich misses 11 fires. Wich makes it very unreliable.
```

# 4. Unit converter - inch to cm

```python
screen_size = int(input("Enter your screen size in inches:"))

screen_size_cm = screen_size * 2.54

print(f"Your screen size in centimeters is {screen_size_cm:.2f} cm.")
```
